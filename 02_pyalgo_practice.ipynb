## Financial Data

```python
url = 'https://certificate.tpq.io/findata.csv'
```

```python
raw = pd.read_csv(url, index_col=0, parse_dates=True)
```

```python
print(raw)
```

```python
raw.info()
```

```python
symbol = 'EUR='
```

```python
data = pd.DataFrame(raw[symbol]).dropna()
```

```python
data.plot();
```

```python
data['r'] = np.log(data / data.shift(1))
```

```python
data.head()
```

```python
data['d'] = np.sign(data['r'])
```

```python
lags = 5
```

```python
cols = list()
for lag in range(1, lags + 1):
    col = f'lag_{lag}'
    data[col] = data['d'].shift(lag)
    # data[col] = data['r'].shift(lag)
    cols.append(col)
```

```python
cols
```

```python
data.head(8)
```

```python
2 ** lags
```

```python
data.dropna(inplace=True)
```

## Strategy Code 

```python
from sklearn.svm import SVC  # classification model
```

```python
from sklearn.neural_network import MLPClassifier  # classification model
```

```python
from sklearn.metrics import accuracy_score
```

```python
# model = GaussianNB()
```

```python
# SVC?
```

```python
# model = SVC(C=1.0, kernel='poly', probability=True)
```

```python
# MLPClassifier?
```

```python
model = MLPClassifier(hidden_layer_sizes=[24, 24], shuffle=False)
```

```python
model.fit(data[cols], data['d'])
```

```python
model.predict(data[cols])
```

```python
data['p'] = model.predict(data[cols])
```

```python
accuracy_score(data['d'], data['p'])
```

```python
model.predict_proba(data[cols])
```

## Backtesting Code

All of the following is calculated under simplifying assumptions, like assuming zero transaction costs and no collateral/margin for short positions.

```python
data['s'] = data['p'] * data['r']
```

```python
data[['r', 's']].sum().apply(np.exp)  # gross returns
```

```python
data[['r', 's']].cumsum().apply(np.exp).plot();  # gross returns over time
```

## Train-Test Split

```python
split = int(len(data) * 0.8)
split
```

```python
train = data.iloc[:split].copy()
```

```python
test = data.iloc[split:].copy()
```

```python
model.fit(train[cols], train['d'])
```

```python
test['p'] = model.predict(test[cols])
```

```python
test['s'] = test['p'] * test['r']
```

```python
test[['r', 's']].sum().apply(np.exp)
```
