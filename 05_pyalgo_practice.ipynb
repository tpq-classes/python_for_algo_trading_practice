## Oanda Demo Account

A demo account with **Oanda UK** is needed to use the REST-API v20 with Oanda. Therefore, you need to register, for example, by using a Virtual Private Network (VPN) Provider to sign up for such an account (e.g. when you are located withing the EU).

The URL is https://oanda.com/uk-en/trading. Then follow the link "Try demo" (to https://oanda.com/apply/demo/).

When you log in to the account (in the "Client Zone"), you land at https://hub.oanda.com. On that page, you find yout **account id** which usually starts with 101-...

On the top left, you see a button "Tools". Click on it and then click on "API". Another page opens where you can create (revoke) an **access token** for the API.

## Historical Data
